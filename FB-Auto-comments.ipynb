```python
"""
group_comment_bot.py
====================
Tu dong binh luan bai viet dau tien trong mot nhom Facebook dinh san.

Thay doi so voi phien ban cu (comment_first_post tren trang chu):
  1. Truy cap truc tiep GROUP_URL thay vi trang chu facebook.com
  2. Doc noi dung bai viet dau tien trong nhom (XPath cap nhat theo cau truc nhom)
  3. Tim o binh luan bang aria-placeholder="Viet binh luan cong khai..."
     (dac trung cua o comment trong nhom, khac trang chu)
  4. Giu nguyen luong tao comment AI (Groq) va gui binh luan

Cach dung:
  - Chay tung cell trong notebook hoac import va goi run_group_commenting(driver, ...)
  - Dam bao driver da dang nhap Facebook truoc khi goi.
"""

from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.keys import Keys
import time
import random
import os


# ============================================================
# CAU HINH - chinh tai day truoc khi chay
# ============================================================

# URL nhom Facebook muon comment bai dau tien
GROUP_URL = "https://www.facebook.com/groups/880359772330567/"  # <- Thay link nhom cua ban

# Groq API key
GROQ_API_KEY = "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"

# Thu muc anh dinh kem vao binh luan (None = khong dinh kem anh)
IMAGE_DIR = r"D:\AIML_Usecases\ai-comment-bot-main\images\vinfast"

# Dinh dang anh duoc chap nhan
IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".gif", ".webp", ".bmp")


# ============================================================
# HELPER
# ============================================================

def random_sleep(min_s: float = 1.0, max_s: float = 3.0):
    """Ngu mot khoang thoi gian ngau nhien (giay)."""
    delay = random.uniform(min_s, max_s)
    time.sleep(delay)


def pick_random_image(image_dir: str) -> str | None:
    """
    Chon ngau nhien mot file anh tu thu muc chi dinh.
    Tra ve duong dan day du (absolute path) hoac None neu khong co anh.
    """
    if not image_dir or not os.path.isdir(image_dir):
        print(f"   WARN Thu muc anh khong hop le: {image_dir}")
        return None

    images = [
        os.path.join(image_dir, f)
        for f in os.listdir(image_dir)
        if f.lower().endswith(IMAGE_EXTENSIONS)
    ]

    if not images:
        print(f"   WARN Khong tim thay anh trong: {image_dir}")
        return None

    chosen = random.choice(images)
    print(f"   OK Chon anh ngau nhien: {os.path.basename(chosen)}")
    return chosen


def attach_image_to_comment(driver, image_path: str) -> bool:
    """
    Dinh kem anh vao o binh luan dang mo.

    Chien luoc: gui duong dan file vao the <input type='file'> an (khong can click button UI).
    Day la cach on dinh nhat voi Selenium vi FB an nut upload bang CSS.

    Tra ve True neu thanh cong, False neu that bai.
    """
    print(f"   Dang dinh kem anh: {os.path.basename(image_path)}...")
    try:
        # Hien thi tat ca input[type=file] an tren trang (FB hay dung display:none)
        driver.execute_script("""
            document.querySelectorAll('input[type="file"]').forEach(function(el) {
                el.style.display = 'block';
                el.style.visibility = 'visible';
                el.style.opacity = '1';
                el.removeAttribute('multiple');  // tranh chon nhieu file
            });
        """)
        random_sleep(0.5, 1.0)

        # Tim input[type=file] lien quan den khu vuc form binh luan
        file_input_xpaths = [
            # Input nam trong form binh luan (chinh xac nhat)
            "//form[@role='presentation']//input[@type='file']",
            # Input co accept anh/video (dac trung cua o dinh kem FB)
            "//input[@type='file' and contains(@accept,'image')]",
            # Fallback: bat ky input file nao tren trang
            "//input[@type='file']",
        ]

        file_input = None
        for xpath in file_input_xpaths:
            try:
                inputs = driver.find_elements(By.XPATH, xpath)
                if inputs:
                    file_input = inputs[0]
                    break
            except Exception:
                continue

        if file_input is None:
            print("   WARN Khong tim thay input[type=file], bo qua buoc dinh kem anh.")
            return False

        # Gui duong dan anh vao input
        file_input.send_keys(image_path)
        print("   OK Da gui duong dan anh vao input file.")

        # Cho preview anh hien ra (thuong mat 2-4 giay)
        random_sleep(3, 5)
        return True

    except Exception as e:
        print(f"   WARN Loi khi dinh kem anh: {e}")
        return False


# ============================================================
# AI - Tao binh luan bang Groq (LLaMA 3)
# ============================================================

def generate_groq_comment(post_content: str) -> str:
    from groq import Groq

    print("Dang nho Groq phan tich noi dung va viet binh luan...")
    client = Groq(api_key=GROQ_API_KEY)

    prompt = f"""
Ban la mot nguoi dung mang xa hoi binh thuong, hay luot Facebook de tro chuyen va chia se kinh nghiem mua sam san sale.

NHIEM VU CUA BAN:
1. Doc noi dung bai viet Facebook duoi day va viet 1 cau dau tien that ngan gon de phan hoi lai noi dung do mot cach dong cam, vui ve, hai huoc va co phan bui bam cuoc song doi thuong.
2. Dung 1 cau that ngan chuyen huong (vi du: "Nhan tien...", "Tien day...", "Noi moi nho...", "San tien ban chuyen nay...") de be lai cau chuyen sang thoi tiet nang nong hoac viec mua sam.
3. Viet 1 cau cuoi review that ngan, chan thuc rang ban vua mua mot chiec ao chong nang rat xin, gia lai re va de lai link: https://www.facebook.com/share/p/1LVizL6JTJ/ keu goi moi nguoi tham khao. Hay dung tu ngu dan da, gan gui nhu "chot don", "ngon bo re", "mac mat ruoi", "xin xo", "ba con tham khao nhe".
link cua san pham se dat o cuoi binh luan.

Vi du ve cau truc: [Binh luan ve bai viet], viet 1 cau theo ngon ngu xue xoa, hai huoc, phan tinh nghich va gan link nay vao cuoi binh luan
https://www.facebook.com/share/p/1LVizL6JTJ/

Noi dung bai viet:
"{post_content}"

YEU CAU QUAN TRONG: Viet lien mach thanh 1 doan van duy nhat, khong xuong dong, khong co cau mo dau chao hoi nhu "Duoi day la binh luan cua toi". Chi in ra noi dung binh luan cuoi cung.
"""

    chat_completion = client.chat.completions.create(
        messages=[{"role": "user", "content": prompt}],
        model="llama-3.3-70b-versatile",
        temperature=0.7,
    )

    comment_text = chat_completion.choices[0].message.content.strip()

    if comment_text.startswith('"') and comment_text.endswith('"'):
        comment_text = comment_text[1:-1]

    print(f"Groq da tao binh luan: {comment_text}")
    return comment_text


# ============================================================
# HAM CHINH - comment bai dau tien trong nhom
# ============================================================

def comment_first_post_in_group(driver, group_url: str = GROUP_URL, image_dir: str = IMAGE_DIR):
    try:
        # 1. Truy cap nhom
        print("1. Truy cap nhom...")
        driver.get(group_url)
        random_sleep(5, 8)

        # 2. Doc noi dung bai viet dau tien trong nhom
        print("2. Dang doc noi dung bai viet dau tien trong nhom...")
        post_content = ""

        post_content_xpaths = [
            (
                "//div[contains(@class,'html-div')]//h3//strong",
                "h3 strong (tieu de bai viet)"
            ),
            (
                "//div[@data-ad-preview='message']//span[@dir='auto']",
                "data-ad-preview message span"
            ),
            (
                "//div[@data-pagelet='GroupFeed']//div[@role='article']//div[@data-ad-comet-preview='message']",
                "GroupFeed article message"
            ),
            (
                "/html/body/div[1]/div/div[1]/div/div[3]/div/div/div[1]/div[1]/div[4]/div/div/div[2]/div/div/div/div[2]/div/div[2]/div/div/div/div/div/div/div/div/div/div[13]/div/div/div[3]/div[1]/div/div",
                "XPath tuyet doi (fallback)"
            ),
        ]

        for xpath, label in post_content_xpaths:
            try:
                post_element = WebDriverWait(driver, 5).until(
                    EC.presence_of_element_located((By.XPATH, xpath))
                )
                post_content = post_element.text.strip()
                if post_content:
                    print(f"   OK Doc noi dung bang [{label}]: {post_content[:120]}...")
                    break
            except Exception:
                print(f"   WARN Khong tim thay bang [{label}], thu cach tiep theo...")

        if not post_content:
            post_content = "Mot bai viet thu vi trong nhom!"
            print("   WARN Khong doc duoc noi dung, dung noi dung mac dinh.")

        random_sleep(1, 2)

        # 3. Tim va click nut "Binh luan" de mo o nhap binh luan
        print("3. Dang tim nut 'Binh luan'...")

        comment_btn_xpaths = [
            # Vietnamese text matching (most common)
            u"//div[@role='button']//span[contains(text(), 'B\u00ecnh lu\u1eadn') or contains(text(), 'Comment')]",
            # aria-label variants
            u"//div[contains(@aria-label,'B\u00ecnh lu\u1eadn') or contains(@aria-label,'Leave a comment') or contains(@aria-label,'Comment')]",
            # Absolute XPath fallback from user-provided HTML
            "/html/body/div[1]/div/div[1]/div/div[3]/div/div/div[1]/div[1]/div[4]/div/div/div[2]/div/div/div/div[2]/div/div[2]/div/div/div/div/div/div/div/div/div/div[13]/div/div/div[4]/div/div/div[1]/div/div[2]/div/div[2]/div/div[1]/div[2]/span",
        ]

        comment_button = None
        for xpath in comment_btn_xpaths:
            try:
                comment_button = WebDriverWait(driver, 5).until(
                    EC.element_to_be_clickable((By.XPATH, xpath))
                )
                print("   OK Tim thay nut Binh luan.")
                break
            except Exception:
                continue

        if comment_button:
            driver.execute_script("arguments[0].scrollIntoView({block: 'center'});", comment_button)
            random_sleep(1, 2)
            driver.execute_script("arguments[0].click();", comment_button)
            random_sleep(1, 2)
        else:
            print("   WARN Khong tim thay nut Binh luan - se thu click thang vao o comment.")

        # 4. Tao binh luan bang AI (Groq)
        ai_comment = generate_groq_comment(post_content)

        # 5. Tim o nhap binh luan va dien noi dung
        print("4. Dang nhap noi dung binh luan...")
        comment_box = None

        comment_box_xpaths = [
            # Priority 1: exact aria-placeholder match for public group comment box (Vietnamese)
            u"//div[@aria-placeholder='Vi\u1ebft b\u00ecnh lu\u1eadn c\u00f4ng khai\u2026' and @contenteditable='true']",
            # Priority 2: English variant
            u"//div[@aria-placeholder='Write a public comment\u2026' and @contenteditable='true']",
            # Priority 3: partial match on Vietnamese 'binh luan'
            u"//div[contains(@aria-placeholder,'b\u00ecnh lu\u1eadn') and @contenteditable='true']",
            # Priority 4: generic contenteditable textbox
  fb_auto_comment_group_vinfastntenteditable='true' and @role='textbox']",
            # Priority 5: absolute XPath from user-provided HTML (last resort)
fb_auto_comment_group_vinfastody/div[1]/div/div[1]/div/div[3]/div/div/div[1]/div[1]/div[4]/div/div/div[2]/div/div/div/div[2]/div/div[2]/div/div/div/div/div/div/div/div/div/div[13]/div/div/div[4]/div/div/div[2]/div[4]/div/div/div/div/div/div[2]/form/div/div/div[1]",
        ]

        for xpath in comment_box_xpaths:
            try:
                comment_box = WebDriverWait(driver, 8).until(
                    EC.presence_of_element_located((By.XPATH, xpath))
                )
                driver.execute_script("arguments[0].scrollIntoView({block: 'center'});", comment_box)
                random_sleep(0.5, 1.5)
                driver.execute_script("arguments[0].click();", comment_box)
                random_sleep(0.5, 1.5)
                print("   OK Tim thay o binh luan, bat dau go...")
                break
            except Exception:
                continue

        if comment_box is None:
            print("   WARN Khong tim thay o binh luan qua XPath, thu go vao active element...")
            comment_box = driver.switch_to.active_element

        comment_box.send_keys(ai_comment)
        random_sleep(1, 2)

        # 6. Dinh kem anh (neu co thu muc anh)
        if image_dir:
            image_path = pick_random_image(image_dir)
            if image_path:
                print("5. Dang dinh kem anh vao binh luan...")
                attach_image_to_comment(driver, image_path)
                # Click lai vao o comment de dam bao focus truoc khi Enter
                try:
                    driver.execute_script("arguments[0].click();", comment_box)
                except Exception:
                    pass
                random_sleep(1, 2)

        # 7. Gui binh luan bang phim Enter
        print("6. Dang gui binh luan...")
        comment_box.send_keys(Keys.RETURN)
        print("Da dang binh luan thanh cong!")
        random_sleep(2, 3)

    except Exception as e:
        print(f"Loi trong qua trinh thao tac: {e}")
        raise


# ============================================================
# VONG LAP - lap lai viec vao nhom va comment bai dau tien
# ============================================================

def run_group_commenting(
    driver,
    group_url: str = GROUP_URL,
    max_comments: int = 10,
    image_dir: str = IMAGE_DIR,
):
    for i in range(max_comments):
        print(f"\n{'='*45}")
        print(f"--- BAT DAU VONG LAP LAN {i+1}/{max_comments} ---")

        try:
            comment_first_post_in_group(driver, group_url=group_url, image_dir=image_dir)

            pause_time = random.uniform(51, 68)
            print(f"Hoan thanh comment lan {i+1}. Tam nghi {int(pause_time)} giay de tranh bi Facebook chan...")
            time.sleep(pause_time)

        except Exception as e:
            print(f"Xay ra loi o vong lap {i+1}: {e}")
            print("Se thu lai o vong lap tiep theo sau 15 giay...")
            time.sleep(15)

    print(f"\n{'='*45}")
    print(f"Da hoan thanh xuat sac {max_comments} vong lap!")
```
